```python
import os
import pandas as pd

# Read the classical results into a Pandas DataFrame
classical_df = pd.read_json('classical_solver_results.json')
gnn_df = pd.read_json('gnn_training_results.json')

# Display the DataFrames
print("\nClassical Solver Results DataFrame:")
display(classical_df)

print("\nGNN Results DataFrame:")
display(gnn_df)

# Create a folder named 'results' if it doesn't exist
os.makedirs('results', exist_ok=True)

# Save the DataFrames as CSV files inside the 'results' folder
classical_df.to_csv('results/classical_solver_results.csv', index=False)
gnn_df.to_csv('results/gnn_training_results.csv', index=False)

print("DataFrames saved as 'results/classical_solver_results.csv' and 'results/gnn_training_results.csv'.")
```
